Servidor de Minecraft controlado desde un panel web.

**Instrucciones:**
1. Ejecuta esta celda
2. Abre la URL que aparece en tu navegador
3. Gestiona tu servidor desde el panel web

**Importante:** Esta celda se mantendrá ejecutándose indefinidamente. No la cierres hasta que termines de usar el panel. Para detener todo, usa el botón ■ (stop).

```python
# ============================================================
#  MineColab Panel — Celda de inicio
#  1. Monta Google Drive
#  2. Instala dependencias
#  3. Clona/actualiza el repositorio
#  4. Inicia el panel web
# ============================================================

# Paso A: Montar Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Paso B: Instalar dependencias Python
import subprocess, os

print("Instalando dependencias...")
subprocess.run(['pip', 'install', '-q',
    'flask', 'flask-socketio', 'flask-cors',
    'mcrcon', 'requests', 'jproperties',
    'PyJWT', 'psutil', 'mcstatus', 'rich',
    'Pillow'
], check=True)
print("Dependencias instaladas.")

# Paso C: Clonar o actualizar el repositorio
REPO_URL = "https://github.com/Danko3104/Minecraft_Server.git"
PANEL_PATH = "/content/minecolab-panel"

if not os.path.exists(PANEL_PATH):
    print("Clonando repositorio...")
    subprocess.run(['git', 'clone', REPO_URL, PANEL_PATH], check=True)
    print("Repositorio clonado.")
else:
    print("Actualizando repositorio...")
    subprocess.run(['git', '-C', PANEL_PATH, 'fetch', 'origin'], check=True)
    subprocess.run(['git', '-C', PANEL_PATH, 'reset', '--hard', 'origin/main'], check=True)
    print("Repositorio actualizado.")

# Paso D: Mostrar último commit para verificar versión
result = subprocess.run(
    ['git', '-C', PANEL_PATH, 'log', '--oneline', '-1'],
    capture_output=True, text=True
)
print(f"Versión: {result.stdout.strip()}")

# Paso E: Agregar al path e iniciar
import sys
sys.path.insert(0, PANEL_PATH)
os.chdir(PANEL_PATH)

from scripts.start import launch
launch()
```
